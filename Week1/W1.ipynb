## Theoretical Questions
- What is a pixel, and what values can it hold in an 8-bit image?

    Pixel (picture element) is a single 2D unit of color in an image. It is the building blocks of images.
    8-bit images refer to how many bits are used to describe a pixel color.

    For Grayscale images:
    - 8-bit images use 8-bits per pixel (256 shades of gray)
    For Color images:
    - 8-bit images use 24-bits per pixel (8 for R, 8 for G and 8 for B)

- How is a grayscale image different from a color (RGB) image in terms of matrix representation?

    RGB makes a singel pixel store 3 different channels of colors making it a 3D matrix whereas the grayscale image has a single channel 2D matrix

- What does the shape (480, 640, 3)  mean for an image array?

    This gets the value of the 4th channel of color in the pixel located at 481st row and the 641st column

- Why does OpenCV use BGR instead of RGB channel ordering?

    Due to historical reasons.

- How is matrix multiplication used in image transformations such as rotation?

    Matrix multiplication changes a matrix by transforming it into a different matrix. With the right matrix chosen different kinds of transformations can be represented.

- What is the role of eigenvalues in computer vision applications?

    It detects features in an image. Like corners/edges for example.
    Other applications like face recognition and compression exist.

- Explain the difference between img[y, x]  and img[x, y]  and why theordering matters.

    Assume img was like:
        Y 0  1
    X 0 [[1, 2]
      1  [3, 4]]

    img[0, 1] = 2
    img[1, 0] = 3

- What is the range of pixel values for a 16-bit image?

    For grayscale 16-bit image:
    2^16 = 65536 (shades of gray)

## Theoretical Questions
- Define convolution in your own words. How is it different from simple multiplication?

    Convolution is different because in the manner in which it multiplies. It slides the same kernel across the matrix and gets some weighted sum of the values.

- What is a kernel/filter? Give two examples and explain what each onedetects.

    This defines the weights that the pixel values are multiplied by to give the result.
    A box blur kernel has the same weight to all positions.
    A Sobel X kernel detects sudden changes across the horizontal line.
    A Sobel Y kernel detects sudden changes across the vertical line.

- What is the difference between SAME and VALID padding?

    SAME: conserves input image size
    VALID: shrinks the image size in the output

- If you convolve a 100x100 image with a 33 kernel (no padding, stride=1) what is the output size?
      
    98x98

- Why does a Sobel X filter detect horizontal edges?

    Soble X looks like: 
    [[-1, 0, 1],
     [-2, 0, 2],
     [-1, 0, 1]]

    It tries to take the difference between the right and left columns to detect a big jump in colors which means there is an edge here

- What happens when all kernel values are identical (e.g., all = 1/9)?

    This is exactly a box blur. It averages out the colors with the nearby one bluring the overall image.

- How does stride affect the size of the output feature map?

    Higher stride --> Lower output image size.

- Why does CNN learn kernels automatically during training?

    To detect increasingly complex features

```python
import cv2
import numpy as np
import matplotlib as plt
```

```python
blur_filter = np.array([
    [1, 1, 1],
    [1, 1, 1],
    [1, 1 ,1]
    ], dtype=np.float32)

blur_filter /= 9
```

```python
blurred = cv2.filter2D(img, -1, blur_filter)
```

```python
showImage([blurred, img])
```

```python
blurred.shape, img.shape
```

```python
sobel_x = np.array([
    [-1, 0, 1],
    [-2, 0, 2],
    [-1, 0 ,1]
    ], dtype=np.float32)

sobel_y = sobel_x.T
```

```python
edge_x = cv2.filter2D(img, -1, sobel_x)
edge_y = cv2.filter2D(img, -1, sobel_y)
```

```python
showImage([edge_x, edge_y])
```

```python
edges = edge_x.copy()
for x in range(edges.shape[0]):
    for y in range(edges.shape[1]):
        for c in range(3):
            edges[x, y, c] = max(edges[x, y, c], edge_y[x,y,c]) 
```

```python
showImage(edges)
```
